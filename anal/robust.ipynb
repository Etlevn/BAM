```python
import os
import pandas as pd
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
ANAL_PATH = os.path.join(BASE_DIR, "anal.csv")
df = pd.read_csv(ANAL_PATH)
```

```python
# Initialize results list
summary = []

# Perform statistical analysis for each metric
for metric in ["MSE", "Amplitude", "PV_Variance"]:
    values = df[metric].dropna()
    mean = values.mean()
    std = values.std()
    cv = std / mean if mean != 0 else np.nan  # Coefficient of Variation

    # 95% Confidence Interval
    ci_low, ci_up = stats.t.interval(
        confidence=0.95, df=len(values) - 1, loc=mean, scale=stats.sem(values)
    )
    ci_width = ci_up - ci_low
    ci_width_ratio = ci_width / mean if mean != 0 else np.nan  # Relative CI width

    summary.append({
        "Metric": metric,
        "Mean": round(mean, 2),
        "StdDev": round(std, 2),
        "CV": round(cv, 3),
        "95% CI Lower": round(ci_low, 2),
        "95% CI Upper": round(ci_up, 2),
        "CI Width / Mean": round(ci_width_ratio, 3)
    })

# Convert result to DataFrame
summary_df = pd.DataFrame(summary)
summary_df
```
